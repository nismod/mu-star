```python
"""
Visualise flood maps processed by the mu-star workflow.
"""
```

```python
import hvplot.xarray
import xarray as xr
from geoviews import WMTS
```

```python
PLOT_KWDS = {
    "width": 1000,
    "height": 600,
    "aspect": "equal",
    "clim": (0,2),
    "cmap": "rainbow",
    "tiles": True,
    "tiles_opts": {"alpha": 0.1},
    "widget_location": "bottom",
}
```

```python
jba = xr.open_zarr("../data/processed/hazard/rp/peril-flood/subperil-fluvial/jba.zarr")
jba
```

```python
plot = jba.inundation.sel(gwl=0.75, rp=100).hvplot(**PLOT_KWDS)
plot
```

```python
swio = xr.open_zarr("../data/processed/hazard/rp/peril-flood/subperil-extratropical-cyclone/swio-rafi.zarr")
swio
```

```python
plot = swio.inundation.sel(case=0, rp=100).hvplot(**PLOT_KWDS)
plot
```
